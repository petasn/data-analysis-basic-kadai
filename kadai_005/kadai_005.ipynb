```python
#インポート
import matplotlib.pyplot as plt
import numpy as np

#4個のグラフを表示する領域を作成する(今回は4個作成なので、6個ではなく4個作成しました)
fig, axes = plt.subplots(2,2,tight_layout=True)

#折れ線グラフ
y0 = [1, 2, -5, 2]
axes[0,0].plot(y0)

#sin関数
x = np.linspace(0, 10, 100)
y = 2 + 2 * np.sin(2 * x)

axes[0,1].plot(x, y, linewidth=2.0)
axes[0,1].set(xlim=(0, 10), xticks=np.arange(0, 10),
       ylim=(-1, 5), yticks=np.arange(-1, 6))

#ヒストグラム
h = np.random .normal(15, 5, 2000)
axes[1,0].hist(h)

# 散布図
np.random.seed(3)
x1 = 4 + np.random.normal(0, 2, 60)
y2 = 4 + np.random.normal(0, 2, len(x1))

axes[1,1].scatter(x1, y2)

#グラフを表示
plt.show()
```
